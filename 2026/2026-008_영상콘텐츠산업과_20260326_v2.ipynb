#### 공연 전후 및 공연기간 요일별 유동인구 분석용 데이터 전처리

```python
print("공연전후 및 공연기간 요일별 유동인구 분석용 데이터 전처리 시작")

def preprocess1(df, gdf=grid50pt_tour):
    df['std_ymd'] = df['std_ymd'].astype('datetime64[ns]')
    df['weekday'] = df['std_ymd'].dt.dayofweek
    df['weeknm'] = df['std_ymd'].dt.day_name('ko_KR')
    df['std_ym'] = df['std_ymd'].astype(str).str[:7]
    gdf = pd.merge(gdf, df, how="left", on="id").dropna()
    return gdf

def gender(df):
    gdf = preprocess1(df, gdf=grid50pt_tour)
    gender_dict = {
        'm_sum': [col for col in gdf.columns if 'v_m_' in col],
        'w_sum': [col for col in gdf.columns if 'v_w_' in col]}
    for key, value in gender_dict.items():
        gdf[key] = gdf.loc[:, value].sum(axis=1) / 24
    keys = [key for key in gender_dict.keys()]
    group = gdf.dissolve(by=['std_ym','weeknm','id'], 
                         aggfunc={keys[0]: np.sum, keys[1]: np.sum}).reset_index()
    group = group.melt(id_vars=['std_ym','weeknm','id','geometry'], value_vars=keys, var_name='gender', value_name='pop')
    group.loc[group['gender'].str.contains('m_'), '성별'] = '남성'
    group.loc[group['gender'].str.contains('w_'), '성별'] = '여성'
    group.rename(columns={'std_ym': '기준연월', 'weeknm': '요일', 'pop': '생활인구'}, inplace=True)
    group = group[['기준연월','요일','id','성별','생활인구','geometry']]
    return group

def age(df):
    gdf = preprocess1(df, gdf=grid50pt_tour)
    ages_dict = {
        '20le_sum': [col for col in df.columns if col in ['v_m_0009','v_m_1014','v_m_1519','v_w_0009','v_w_1014','v_w_1519']], 
        '20eq_sum': [col for col in df.columns if col in ['v_m_2024','v_m_2529','v_w_2024','v_w_2529']],
        '30eq_sum': [col for col in df.columns if col in ['v_m_3034','v_m_3539','v_w_3034','v_w_3539']], 
        '40eq_sum': [col for col in df.columns if col in ['v_m_4044','v_m_4549','v_w_4044','v_w_4549']],
        '50eq_sum': [col for col in df.columns if col in ['v_m_5054','v_m_5559','v_w_5054','v_w_5559']], 
        '60gt_sum': [col for col in df.columns if col in ['v_m_6064','v_m_6569','v_m_7000','v_w_6064','v_w_6569','v_w_7000']]}
    for key, value in ages_dict.items():
        gdf[key] = gdf.loc[:, value].sum(axis=1) / 24
    keys = [key for key in ages_dict.keys()]
    group = gdf.dissolve(by=['std_ym','weeknm','id'], 
                         aggfunc={keys[0]: np.sum, keys[1]: np.sum, keys[2]: np.sum, 
                                  keys[3]: np.sum, keys[4]: np.sum, keys[5]: np.sum}).reset_index()
    group = group.melt(id_vars=['std_ym','weeknm','id','geometry'], value_vars=keys, var_name='age', value_name='pop')
    group.loc[group['age'].str.startswith('20le'), '연령'] = '20대미만'
    group.loc[group['age'].str.startswith('20eq'), '연령'] = '20대'
    group.loc[group['age'].str.startswith('30eq'), '연령'] = '30대'
    group.loc[group['age'].str.startswith('40eq'), '연령'] = '40대'
    group.loc[group['age'].str.startswith('50eq'), '연령'] = '50대'
    group.loc[group['age'].str.startswith('60gt'), '연령'] = '60대이상'
    group.rename(columns={'std_ym': '기준연월', 'weeknm': '요일', 'pop': '생활인구'}, inplace=True)
    group = group[['기준연월','요일','id','연령','생활인구','geometry']]
    return group

def time(df):
    gdf = preprocess1(df, gdf=grid50pt_tour)
    time_dict = dict((f'{h:02d}H', [f'v_t_{h:02d}']) for h in range(24))
    for key, value in time_dict.items():
        gdf[key] = gdf.loc[:, value].sum(axis=1)
    keys = [key for key in time_dict.keys()]
    group = gdf.dissolve(by=['std_ym','weeknm','id'],
                       aggfunc={keys[0]: np.sum, keys[1]: np.sum, keys[2]: np.sum, keys[3]: np.sum, keys[4]: np.sum,
                                keys[5]: np.sum, keys[6]: np.sum, keys[7]: np.sum, keys[8]: np.sum, keys[9]: np.sum,
                                keys[10]: np.sum, keys[11]: np.sum, keys[12]: np.sum, keys[13]: np.sum, keys[14]: np.sum,
                                keys[15]: np.sum, keys[16]: np.sum, keys[17]: np.sum, keys[18]: np.sum, keys[19]: np.sum,
                                keys[20]: np.sum, keys[21]: np.sum, keys[22]: np.sum, keys[23]: np.sum}).reset_index()
    group = group.melt(id_vars=['std_ym','weeknm','id','geometry'], value_vars=keys, var_name='time', value_name='pop')
    group.rename(columns={'std_ym': '기준연월', 'weeknm': '요일', 'time': '시간', 'pop': '생활인구'}, inplace=True)
    group = group[['기준연월','요일','id','시간','생활인구','geometry']]
    return group

genders = gender(dfs_pop)
ages = age(dfs_pop)
times = time(dfs_tme)
gender(dfs_pop).to_file(data_path1 + "2026-008_데이터 분석 컨설팅 데이터.gpkg", layer="월별_요일_성별_인구현황", driver="GPKG")
pd.DataFrame(gender(dfs_pop)).drop(columns=['geometry']).to_csv(data_path4 + "월별_요일_성별_인구현황.csv", sep="|", encoding="utf-8", index=False)
age(dfs_pop).to_file(data_path1 + "2026-008_데이터 분석 컨설팅 데이터.gpkg", layer="월별_요일_연령_인구현황", driver="GPKG")
pd.DataFrame(age(dfs_pop)).drop(columns=['geometry']).to_csv(data_path4+ "월별_요일_연령_인구현황.csv", sep="|", encoding="utf-8", index=False)
time(dfs_tme).to_file(data_path1 + "2026-008_데이터 분석 컨설팅 데이터.gpkg", layer="월별_요일_시간_인구현황", driver="GPKG")
pd.DataFrame(age(dfs_tme)).drop(columns=['geometry']).to_csv(data_path4+ "월별_요일_시간_인구현황.csv", sep="|", encoding="utf-8", index=False)

print("공연전후 및 공연기간 요일별 유동인구 분석용 데이터 전처리 완료")
```

#### 전년동기 대비 공연기간 월별 인구변화

```python

times['기준연'] = times['기준연월'].str[:4] + "년"
times['기준월'] = times['기준연월'].str[5:7] + "월"
times1 = times[times['기준연월'].str.contains('2024-11|2024-12|2025-01|2025-02|2025-03')].copy()
times2 = times[times['기준연월'].str.contains('2025-11|2025-12|2026-01|2026-02|2026-03')].copy()

time1 = times1.groupby(['기준연','기준월'], as_index=False)[['생활인구']].sum().round().sort_values(['기준연','기준월'], ascending=True)
time1.rename(columns={'기준연': '전년동기_기준연', '생활인구': '전년동기_생활인구'}, inplace=True)
time1['전년동기_평균_생활인구'] = (time1['전년동기_생활인구'] / 24).round()
time2 = times2.groupby(['기준연','기준월'], as_index=False)[['생활인구']].sum().round().sort_values(['기준연','기준월'], ascending=True)
time2.rename(columns={'기준연': '공연기간_기준연', '생활인구': '공연기간_생활인구'}, inplace=True)
time2['공연기간_평균_생활인구'] = (time2['공연기간_생활인구'] / 24).round()

results1 = time1.merge(time2, how="left", on="기준월")
results1['증감량'] = results1['공연기간_평균_생활인구'] - results1['전년동기_평균_생활인구']
print(f"월별 전년동기: ", results1[['전년동기_기준연','기준월','전년동기_평균_생활인구']])
print(f"월별 전년동기 평균: ", results1['전년동기_평균_생활인구'].mean())
print(f"월별 공연기간: ", results1[['공연기간_기준연','기준월','공연기간_평균_생활인구']])
print(f"월별 공연기간 평균: ", results1['공연기간_평균_생활인구'].mean())
# print((results1['전년동기_생활인구'].mean()/24), (results1['공연기간_생활인구'].mean()/24), sep="\n")
print(((results1['공연기간_평균_생활인구'].mean()/24) - (results1['전년동기_평균_생활인구'].mean()/24)) / (results1['전년동기_평균_생활인구'].mean()/24) * 100)
```

```python
print("방문객 평균 체류시간(1시간) 및 이동패턴 분석 시작")

import pandas as pd
import numpy as np
import geopandas as gpd
from scipy.signal import correlate
from pathlib import Path

OUT = Path(data_path4)
OUT.mkdir(exist_ok=True)
GPKG = data_path1 + "2026-008_데이터 분석 컨설팅 데이터.gpkg"

공연기간 = ('2025-11-01', '2026-03-31')
전년동기 = ('2024-11-01', '2025-03-31')
TOURS = ['스페이스원지','영도대교','흰여울문화마을','피아크(아르뗌)']

dfs = dfs_tme.copy()

# 일별 격자별 생활인구(주거+직장+방문) long형 변환
dfs['std_ymd'] = dfs['std_ymd'].astype('datetime64[ns]')
dfs['weekday'] = dfs['std_ymd'].dt.dayofweek
# dfs['std_ymd'] = dfs['std_ymd'].astype(str)
time_dict = dict((f'{h:02d}H', [f'h_t_{h:02d}', f'w_t_{h:02d}', f'v_t_{h:02d}']) for h in range(0, 24))
for key, value in time_dict.items():
    dfs[key] = dfs.loc[:, value].sum(axis=1).round()
keys = [key for key in time_dict.keys()]
dfs = dfs.melt(id_vars=['std_ymd','id','weekday'], value_vars=keys, var_name='hours', value_name='pop')
dfs['hour'] = dfs['hours'].str.extract(r'(\d+)').astype(int)
id_to_tour = grid50_tour[['id','tourname']]
dfs = dfs.merge(id_to_tour, on='id', how="inner").dropna()
dfs['tourname'] = dfs['tourname'].astype('category')

# 일별 시간대 체류인구 생성
stay = dfs.groupby(['std_ymd','tourname','hour'], observed=True)['pop'].sum().reset_index()
stay['weekday'] = stay['std_ymd'].dt.dayofweek

# 평균 체류시간(+기간비교)
stay_daily = stay.groupby(['std_ymd','tourname'], observed=True).agg(
    person_hours=('pop', 'sum'), peak=('pop', 'max')
    ).reset_index()
stay_daily['avg_stay_hour'] = stay_daily['person_hours'] / stay_daily['peak']
stay_daily['weekday'] = stay_daily['std_ymd'].dt.dayofweek

def period_filter(df_in, start, end, weekdays=None):
    m = (df_in['std_ymd'] >= start) & (df_in['std_ymd'] <= end)
    if weekdays is not None:
        m &= df_in['weekday'].isin(weekdays)
    return df_in[m]

print("4-1. [신청 3-B] 관광지별 체류인구 변화 (주말/주중)")
def tour_avg(df_in, start, end, weekdays):
    return period_filter(df_in, start, end, weekdays) \
        .groupby('tourname')['person_hours'].mean()

rows = []
for label, days in [('주말', [5,6]), ('주중', [0,1,2,3,4])]:
    prev = tour_avg(stay_daily, *전년동기, days)
    curr = tour_avg(stay_daily, *공연기간, days)
    for tour in TOURS:
        if tour not in prev.index or tour not in curr.index:
            continue
        p, c = prev[tour], curr[tour]
        rows.append({
            '기간구분': label,
            '관광지': tour,
            '전년동기_인시간': round(p),
            '공연기간_인시간': round(c),
            '증감률(%)': round((c/p - 1) * 100, 2)
        })
report_3b = pd.DataFrame(rows)

print("4-2. [신청 3-C] 시간대별 체류 패턴 (공연기간 주말)")
sat_show = period_filter(stay, *공연기간, [5,6])
hourly_pattern = sat_show.pivot_table(
    index='hour', columns='tourname', values='pop', aggfunc='mean'
).round(0)

peak_hour = hourly_pattern.idxmax().reset_index()
peak_hour.columns = ['관광지','피크시간']

print("4-3. 이동패턴 (1) 시차 교차상관")
records = []
for origin in TOURS:
    for dest in TOURS:
        if origin == dest: continue
        if origin not in hourly_pattern.columns or dest not in hourly_pattern.columns:
            continue
        o, d = hourly_pattern[origin].values, hourly_pattern[dest].values
        do, dd = np.diff(o), np.diff(d)
        xcorr = correlate(dd, -do, mode='full')
        lags = np.arange(-len(do)+1, len(do))
        best_lag = int(lags[np.argmax(xcorr)])
        records.append({
            '출발지': origin,
            '목적지': dest,
            '최적시차(h)': best_lag,
            '상관강도': round(float(xcorr.max()), 1),
            '해석': (
                f'{origin} 감소 후 {best_lag}시간 뒤 {dest} 증가' if best_lag > 0
                else (f'동시 변동' if best_lag == 0
                      else f'{dest} 감소 후 {-best_lag}시간 뒤 {origin} 증가')
            )
        })
lag_result = pd.DataFrame(records)

print("4-4. 이동패턴 (2) 동시방문 (Z-score)")
daily_pivot = stay.groupby(['std_ymd','tourname'], observed=True)['pop'] \
                  .sum().unstack('tourname')

show_pivot = daily_pivot.loc[공연기간[0]:공연기간[1]]
z = (show_pivot - show_pivot.mean()) / show_pivot.std()

idx = show_pivot.index
sat_mask = idx.dayofweek == 5
sun_mask = idx.dayofweek == 6

co_records = []
for a in TOURS:
    for b in TOURS:
        if a >= b: continue
        if a not in z.columns or b not in z.columns: continue
        co_sat = ((z.loc[sat_mask, a]>0) & (z.loc[sat_mask, b]>0)).mean()
        co_sun = ((z.loc[sun_mask, a]>0) & (z.loc[sun_mask, b]>0)).mean()
        co_records.append({
            '관광지A': a, '관광지B': b,
            '토요일_동시방문률(%)': round(co_sat*100, 1),
            '일요일_동시방문률(%)': round(co_sun*100, 1),
            '차이(토-일,%p)': round((co_sat - co_sun)*100, 1)
        })
covisit = pd.DataFrame(co_records)

print("4-5. QNEAT3 동선 (이동패턴 보조)")
v1 = gpd.read_file(GPKG, layer='visitor_route1')
v2 = gpd.read_file(GPKG, layer='visitor_route2')

route_segments = []
for label, gdf in [('동선1(최단)', v1), ('동선2(대안)', v2)]:
    for _, r in gdf.iterrows():
        route_segments.append({
            '동선': label,
            '구간': r['path'],
            '거리(km)': round(r['length']/1000, 2)
        })
route_detail = pd.DataFrame(route_segments)

route_summary = pd.DataFrame([
    {'동선': '동선1(최단)',
     '총거리(km)': round(v1['length'].sum()/1000, 2),
     '구간수': len(v1),
     '경로': ' → '.join(v1['path'].tolist())},
    {'동선': '동선2(대안)',
     '총거리(km)': round(v2['length'].sum()/1000, 2),
     '구간수': len(v2),
     '경로': ' → '.join(v2['path'].tolist())}
])

print("4-6. 보고서용 핵심 KPI 요약 (시트 0)")
def safe_get(df_in, query_str, col):
    try:
        v = df_in.query(query_str)[col]
        return v.iloc[0] if len(v) > 0 else None
    except Exception:
        return None

kpi_summary = pd.DataFrame([
    {'지표': '[3-B] 주말 영도대교 체류 증감률(%)',
     '값': safe_get(report_3b, "기간구분=='주말' and 관광지=='영도대교'", '증감률(%)')},
    {'지표': '[3-B] 주말 흰여울문화마을 체류 증감률(%)',
     '값': safe_get(report_3b, "기간구분=='주말' and 관광지=='흰여울문화마을'", '증감률(%)')},
    {'지표': '[3-B] 주말 스페이스원지 체류 증감률(%)',
     '값': safe_get(report_3b, "기간구분=='주말' and 관광지=='스페이스원지'", '증감률(%)')},
    {'지표': '[3-B] 주말 피아크,아르뗌 체류 증감률(%)',
     '값': safe_get(report_3b, "기간구분=='주말' and 관광지=='피아크, 아르뗌'", '증감률(%)')},
    {'지표': '[3-C] 스페이스원지 피크 시간',
     '값': safe_get(peak_hour, "관광지=='스페이스원지'", '피크시간')},
    {'지표': '[3-C] 영도대교 피크 시간',
     '값': safe_get(peak_hour, "관광지=='영도대교'", '피크시간')},
    {'지표': '[이동] 토-일 동시방문률 차이(스페이스원지-영도대교, %p)',
     '값': safe_get(covisit, "관광지A=='스페이스원지' and 관광지B=='영도대교'", '차이(토-일,%p)')},
    {'지표': '[이동] QNEAT3 동선1 총거리(km)',
     '값': route_summary.iloc[0]['총거리(km)']},
    {'지표': '[이동] QNEAT3 동선2 총거리(km)',
     '값': route_summary.iloc[1]['총거리(km)']},
])

print("4-7. 통합 엑셀 저장")
output_path = OUT/'step4_분석결과.xlsx'

with pd.ExcelWriter(output_path, engine='openpyxl') as w:
    kpi_summary.to_excel(w, sheet_name='0_KPI요약', index=False)
    report_3b.to_excel(w, sheet_name='1_3B_체류변화', index=False)
    hourly_pattern.to_excel(w, sheet_name='2_3C_시간대패턴')
    peak_hour.to_excel(w, sheet_name='3_3C_피크시간', index=False)
    lag_result.to_excel(w, sheet_name='4_시차교차상관', index=False)
    z.round(3).to_excel(w, sheet_name='5_Zscore일별')
    covisit.to_excel(w, sheet_name='6_동시방문', index=False)
    route_summary.to_excel(w, sheet_name='7_QNEAT3동선요약', index=False)
    route_detail.to_excel(w, sheet_name='8_QNEAT3동선상세', index=False)

# 열 너비 자동 조정 (가독성 향상)
from openpyxl import load_workbook
wb = load_workbook(output_path)
for sheet_name in wb.sheetnames:
    ws = wb[sheet_name]
    for col in ws.columns:
        max_len = 0
        col_letter = col[0].column_letter
        for cell in col:
            try:
                if cell.value is not None:
                    max_len = max(max_len, len(str(cell.value)))
            except: pass
        ws.column_dimensions[col_letter].width = min(max_len + 2, 40)
wb.save(output_path)

print(f'저장 완료: {output_path.resolve()}')
print(f'시트 수: {len(wb.sheetnames)}')
print(f'시트 목록: {wb.sheetnames}')

print("방문객 평균 체류시간(1시간) 및 이동패턴 분석 완료")
```

```python
# !pip install koreanize_matplotlib
# 방문객 평균 체류시간(1시간) 및 이동패턴 분석
import pandas as pd
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

time1 = inputs[0].drop(columns=['dat_flow_id', 'part_batchdate'])
time2 = inputs[1].drop(columns=['dat_flow_id', 'part_batchdate'])
grdlists = inputs[2]

cols = [col for col in time1.columns if '_t_' in col]
time1[cols] = time1[cols].astype(float)
time = pd.concat([time1, time2], ignore_index=True)
results = tour50m.merge(time, how="left", on="id").dropna()

DEPARTURE = '스페이스원지'
DESTINATIONS = ['영도대교', '흰여울문화마을', '피아크, 아르뗌']
ALL_POIS = [DEPARTURE] + DESTINATIONS


# ═════ STEP 2: 방문성 지수 + 임계값 ═════
def add_visit_ratio(df, day_hours_only=True):
    """방문성 지수 계산 (주간 09~21시 기본)."""
    out = df.copy()
    if day_hours_only:
        h_cols = [f'h_t_{h:02d}' for h in range(9, 22)]
        w_cols = [f'w_t_{h:02d}' for h in range(9, 22)]
        v_cols = [f'v_t_{h:02d}' for h in range(9, 22)]
        n = 13
    else:
        h_cols = [f'h_t_{h:02d}' for h in range(24)]
        w_cols = [f'w_t_{h:02d}' for h in range(24)]
        v_cols = [f'v_t_{h:02d}' for h in range(24)]
        n = 24
    out['H'] = out[h_cols].sum(axis=1) / n
    out['W'] = out[w_cols].sum(axis=1) / n
    out['V'] = out[v_cols].sum(axis=1) / n
    out['visit_ratio'] = out['V'] / (out['H'] + out['W'] + out['V'])
    return out


def filter_grids(res, ratio_col='visit_ratio', tour_col='tourname',
                 abs_thresh='auto', rel_thresh=0.25):
    """절대 + 관광지 내부 상대 임계값 적용.

    abs_thresh='auto': 격자별 평균 visit_ratio 분포의 P25를 자동 적용
                       (격자 단위 분포 기준이므로 day_hours_only 변경에도 적응)
    abs_thresh=숫자  : 명시적 임계값
    abs_thresh=None  : 절대 임계값 미적용 (상대 임계값만 사용)
    """
    df = res.copy()

    if abs_thresh == 'auto':
        # 격자별 평균 visit_ratio로 P25 산출 (날짜 변동 제거)
        grid_mean = df.groupby('id')[ratio_col].mean()
        abs_thresh = grid_mean.quantile(0.25)
        print(f"  자동 절대 임계값 (격자별 평균 P25): {abs_thresh:.4f}")
    elif abs_thresh is None:
        abs_thresh = -np.inf

    df['rank_within_tour'] = df.groupby(tour_col)[ratio_col].rank(pct=True)
    cond = (df[ratio_col] >= abs_thresh) & (df['rank_within_tour'] >= rel_thresh)
    filtered = df[cond].copy()

    print(f"\n[STEP 2] 임계값 필터링 (abs={abs_thresh:.3f}, rel={rel_thresh})")
    print(f"  전체 id: {df['id'].nunique()} → 채택: {filtered['id'].nunique()}")
    print(filtered.groupby(tour_col)['id'].nunique()
          .rename('채택 id 수').to_frame())

    if filtered['id'].nunique() == df['id'].nunique():
        print("  ⚠ 필터링이 작동하지 않음 — 임계값을 더 높이거나 rel_thresh 상향 필요")
    return filtered


# ═════ 1시간 이상 체류 신호 ═════
def add_long_stay_signal(df, hour_range=(6, 23)):
    """ls_h = min(v_t_h, v_t_h+1)."""
    h0, h1 = hour_range
    out = df.copy()
    for h in range(h0, h1):
        c1, c2 = f'v_t_{h:02d}', f'v_t_{h+1:02d}'
        out[f'ls_{h:02d}'] = df[[c1, c2]].min(axis=1)
    return out


# ═════ STEP 3: 인구현황 ═════
def population_status_monthly(df, tour_col='tourname', hour_range=(6, 23),
                               output='wide'):
    df = df.copy()
    df['date'] = pd.to_datetime(df['std_ymd'], format='%Y%m%d')
    df['year_month'] = df['date'].dt.strftime('%Y-%m')
    df['day_type'] = np.where(df['date'].dt.weekday >= 5, 'weekend', 'weekday')

    h0, h1 = hour_range
    ls_cols = [f'ls_{h:02d}' for h in range(h0, h1)]
    daily = (df.groupby([tour_col, 'year_month', 'day_type', 'std_ymd'])
             [ls_cols].sum())
    monthly = daily.groupby(level=[tour_col, 'year_month', 'day_type']).mean().round(1)

    if output == 'wide':
        return monthly
    long = monthly.reset_index().melt(
        id_vars=[tour_col, 'year_month', 'day_type'],
        value_vars=ls_cols, var_name='hour_col', value_name='체류인구',
    )
    long['시간대'] = long['hour_col'].str.extract(r'ls_(\d+)').astype(int)
    long = long.rename(columns={
        tour_col: '관광지', 'year_month': '연월', 'day_type': '요일유형',
    })[['관광지', '연월', '요일유형', '시간대', '체류인구']]
    return long.sort_values(['관광지', '연월', '요일유형', '시간대'])


def population_profile(df, tour_col='tourname', hour_range=(6, 23),
                       date_filter=None):
    """평일/주말 평균 프로파일.

    date_filter: ('YYYYMMDD', 'YYYYMMDD') 형식 튜플로 기간 한정 가능
    """
    df = df.copy()
    if date_filter is not None:
        start, end = date_filter
        mask = (df['std_ymd'].astype(str) >= str(start)) & \
               (df['std_ymd'].astype(str) <= str(end))
        df = df[mask]

    df['weekday'] = pd.to_datetime(df['std_ymd'], format='%Y%m%d').dt.weekday
    df['day_type'] = np.where(df['weekday'] >= 5, 'weekend', 'weekday')

    h0, h1 = hour_range
    ls_cols = [f'ls_{h:02d}' for h in range(h0, h1)]
    daily = df.groupby([tour_col, 'std_ymd', 'day_type'])[ls_cols].sum()
    return daily.groupby(level=[tour_col, 'day_type']).mean()


# ═════ STEP 4: 장기체류지수 (수정: max 비율 방식) ═════
def long_stay_index(profile, min_consecutive=2, max_ratio=0.5):
    """관광지별 체류형 vs 통과형 분류.

    임계값 = 그 관광지 v 곡선의 최댓값 × max_ratio
    핵심 체류 시간대 = 임계값 초과 + 연속 min_consecutive 시간 이상
    """
    rows = []
    for (poi, day_type), series in profile.iterrows():
        v_arr = series.values
        hours = [int(c.replace('ls_', '')) for c in series.index]
        total = v_arr.sum()
        if total == 0:
            continue

        threshold = np.max(v_arr) * max_ratio
        above = v_arr >= threshold

        long_mask = np.zeros_like(above)
        i = 0
        intervals = []
        while i < len(above):
            if above[i]:
                j = i
                while j < len(above) and above[j]:
                    j += 1
                if j - i >= min_consecutive:
                    long_mask[i:j] = True
                    intervals.append((hours[i], hours[j-1]))
                i = j
            else:
                i += 1

        ls_idx = v_arr[long_mask].sum() / total
        peak_h = hours[np.argmax(v_arr)]
        if ls_idx >= 0.6:
            vtype = '체류형'
        elif ls_idx >= 0.3:
            vtype = '혼합형'
        else:
            vtype = '통과형'
        core_str = ', '.join(f'{a}-{b+1}시' for a, b in intervals) if intervals else '없음'

        rows.append({
            'poi': poi, 'day_type': day_type,
            'long_stay_index': round(ls_idx, 3),
            'peak_hour': peak_h, 'core_hours': core_str, 'venue_type': vtype,
        })
    return pd.DataFrame(rows).set_index(['poi', 'day_type'])


def report_long_stay_index(profile, max_ratio=0.5):
    print("\n" + "=" * 72)
    print(f"[STEP 4] 장기체류지수 (임계값 = max × {max_ratio})")
    print("=" * 72)
    ls_df = long_stay_index(profile, max_ratio=max_ratio)
    print(ls_df.to_string())
    print("\n  ≥0.6: 체류형 / 0.3~0.6: 혼합형 / <0.3: 통과형")
    return ls_df


# ═════ STEP 5: 이동패턴 (수정: 동기화 vs 흐름 구분) ═════
def normalize_profile(profile, day_type='weekend'):
    sub = profile.xs(day_type, level='day_type')
    max_v = sub.max(axis=1).replace(0, np.nan)
    return sub.div(max_v, axis=0)


def flow_analysis(profile_norm, max_lag=3):
    if DEPARTURE not in profile_norm.index:
        raise ValueError(f"출발지 '{DEPARTURE}' 없음")

    n = profile_norm.shape[1]
    v_dep = profile_norm.loc[DEPARTURE].values
    dv_dep = np.diff(v_dep)

    peak_h = profile_norm.idxmax(axis=1).str.extract(r'ls_(\d+)')[0].astype(int)
    sp_peak = peak_h[DEPARTURE]

    rows = []
    for dest in DESTINATIONS:
        if dest not in profile_norm.index:
            continue
        v_arr = profile_norm.loc[dest].values
        dv_arr = np.diff(v_arr)

        # 동기 상관 (τ=0) 추가 — 동기화 vs 흐름 구분 핵심
        rho_sync = np.corrcoef(v_dep, v_arr)[0, 1]

        best = {'tau': None, 'rho_fwd': -1, 'rho_bwd': 0, 'rho_d': -1}
        for tau in range(1, max_lag + 1):
            r_fwd = np.corrcoef(v_dep[:n-tau], v_arr[tau:])[0, 1]
            r_bwd = np.corrcoef(v_arr[:n-tau], v_dep[tau:])[0, 1]
            x = -dv_dep[:len(dv_dep)-tau+1]
            y = dv_arr[tau-1:]
            if len(x) > 1 and np.std(x) > 0 and np.std(y) > 0:
                r_d = np.corrcoef(x, y)[0, 1]
            else:
                r_d = np.nan
            if r_fwd > best['rho_fwd']:
                best = {'tau': tau, 'rho_fwd': r_fwd, 'rho_bwd': r_bwd,
                        'rho_d': r_d}

        rows.append({
            'destination': dest,
            'peak_lag_h': peak_h[dest] - sp_peak,
            'rho_sync': rho_sync,
            'tau_h': best['tau'],
            'rho_fwd': best['rho_fwd'],
            'rho_bwd': best['rho_bwd'],
            'asymmetry': best['rho_fwd'] - best['rho_bwd'],
            'rho_derivative': best['rho_d'],
        })
    return pd.DataFrame(rows), sp_peak


def classify_relationship(r):
    """관광지 간 관계 분류: 흐름 / 동기화 / 무관 / 역흐름.

    - 흐름:  피크차 ≥ +1h, 비대칭 ≥ +0.10, ρ_d ≥ +0.30
    - 동기화: 피크차 ≈ 0, ρ_sync ≥ +0.7, 비대칭 ≈ 0
    - 역흐름: 피크차 ≤ -1h
    - 무관:  ρ_sync < 0.3
    """
    if r['rho_sync'] < 0.3:
        return '무관'
    if r['peak_lag_h'] <= -1:
        return '역흐름'
    if r['peak_lag_h'] == 0 and abs(r['asymmetry']) < 0.05 and r['rho_sync'] >= 0.7:
        return '동기화'
    flow_score = ((r['peak_lag_h'] >= 1) +
                  (r['asymmetry'] >= 0.10) +
                  (r['rho_derivative'] >= 0.30))
    if flow_score >= 2:
        return '흐름'
    return '동기화' if r['rho_sync'] >= 0.7 else '약한관계'


def report_flow(profile, day_types=('weekday', 'weekend'), max_lag=3):
    print("\n" + "=" * 80)
    print(f"[STEP 5] {DEPARTURE} → 도착지 관계 분석")
    print("=" * 80)

    all_results = []
    for dt in day_types:
        try:
            pn = normalize_profile(profile, day_type=dt)
            flow, sp_peak = flow_analysis(pn, max_lag=max_lag)
        except (ValueError, KeyError) as e:
            print(f"\n[{dt}] 분석 불가: {e}")
            continue

        print(f"\n■ {dt.upper()}  (스페이스원지 피크: {sp_peak}시)")
        print("-" * 80)
        flow['day_type'] = dt
        flow['sp_peak_hour'] = sp_peak
        relationships = []
        for _, r in flow.iterrows():
            rel = classify_relationship(r)
            relationships.append(rel)
            print(f"  → {r['destination']:18s} "
                  f"피크차={r['peak_lag_h']:+d}h "
                  f"ρ_sync={r['rho_sync']:+.2f} "
                  f"ρ_fwd={r['rho_fwd']:+.2f} "
                  f"비대칭={r['asymmetry']:+.2f} "
                  f"ρ_d={r['rho_derivative']:+.2f}  [{rel}]")
        flow['관계'] = relationships
        all_results.append(flow)

    print("\n[관계 유형 정의]")
    print("  흐름:    SP에서 도착지로 시간차 이동 (피크차+, 비대칭+, ρ_d+)")
    print("  동기화:  같은 시간대에 함께 변동 (피크차=0, ρ_sync 매우 높음)")
    print("  역흐름:  도착지가 SP보다 먼저 (피크차 음수)")
    print("  무관:    상관 거의 없음")

    if all_results:
        return pd.concat(all_results, ignore_index=True)
    return pd.DataFrame()


# ═════ 분기별 흐름 비교 ═════
def quarterly_flow_comparison(df, tour_col='tourname', hour_range=(6, 23)):
    """기간을 분기로 나눠 흐름 패턴이 변화하는지 확인.

    스페이스원지 개장 효과 등을 잡아내는 데 유용.
    """
    df = df.copy()
    df['date'] = pd.to_datetime(df['std_ymd'], format='%Y%m%d')
    df['quarter'] = df['date'].dt.to_period('Q').astype(str)

    print("\n" + "=" * 72)
    print("분기별 흐름 패턴 비교")
    print("=" * 72)

    results = []
    for q, sub in df.groupby('quarter'):
        n_days = sub['std_ymd'].nunique()
        if n_days < 30:
            continue
        try:
            profile = population_profile(sub, tour_col=tour_col,
                                          hour_range=hour_range)
            pn = normalize_profile(profile, day_type='weekend')
            flow, sp_peak = flow_analysis(pn)
            flow['quarter'] = q
            flow['n_days'] = n_days
            results.append(flow)
            print(f"\n■ {q} ({n_days}일 / 스페이스원지 피크 {sp_peak}시)")
            for _, r in flow.iterrows():
                rel = classify_relationship(r)
                print(f"  → {r['destination']:18s} "
                      f"피크차={r['peak_lag_h']:+d}h "
                      f"ρ_sync={r['rho_sync']:+.2f} "
                      f"ρ_fwd={r['rho_fwd']:+.2f} [{rel}]")
        except Exception as e:
            print(f"\n■ {q}: 분석 실패 ({e})")
            continue

    if results:
        return pd.concat(results, ignore_index=True)
    return pd.DataFrame()


# ═════ 시각화 ═════
def plot_population(profile, save_path=None):
    fig, axes = plt.subplots(1, 2, figsize=(15, 5))
    colors = {'스페이스원지': 'red', '영도대교': 'orange',
              '흰여울문화마을': 'blue', '피아크, 아르뗌': 'green'}
    for ax, dt in zip(axes, ['weekday', 'weekend']):
        try:
            sub = profile.xs(dt, level='day_type')
        except KeyError:
            continue
        hours = [int(c.replace('ls_', '')) for c in sub.columns]
        for poi in sub.index:
            lw = 2.5 if poi == DEPARTURE else 1.5
            ax.plot(hours, sub.loc[poi].values, label=poi,
                    color=colors.get(poi, 'gray'), linewidth=lw,
                    marker='o' if poi == DEPARTURE else None, markersize=5)
        ax.set_xlabel('시간 (시)')
        ax.set_ylabel('1시간 이상 체류 인구')
        ax.set_title(f'{dt.upper()} - 시간대별 체류 인구')
        ax.legend(fontsize=9)
        ax.grid(True, alpha=0.3)
    plt.tight_layout()
    if save_path:
        plt.savefig(save_path, dpi=120, bbox_inches='tight')
    plt.show()

# 1) 방문성 지수 + 임계값 (자동 조정)
res2 = add_visit_ratio(results, day_hours_only=True)
filtered = filter_grids(res2, abs_thresh='auto', rel_thresh=0.25)

# 2) 1시간 이상 체류 신호
filtered_ls = add_long_stay_signal(filtered, hour_range=(6, 23))

# 3) 평일/주말 프로파일 + 시각화
profile = population_profile(filtered_ls, hour_range=(6, 23))
plot_population(profile)

# 4) STEP 4: 장기체류지수 (max 비율 방식)
ls_df = report_long_stay_index(profile, max_ratio=0.5)

# 5) STEP 5: 관계 분석 (동기화 vs 흐름 구분)
flow_df = report_flow(profile)

# 6) 분기별 흐름 변화 (선택)
quarterly_df = quarterly_flow_comparison(filtered_ls)
```
