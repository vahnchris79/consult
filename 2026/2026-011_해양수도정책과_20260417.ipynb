```python
import glob
import pandas as pd

data_path = r"D:\02_사업관리\적재확인\02_카드이용\\"

data_lists = glob.glob(data_path + "*.csv", recursive=True)

res = []
for data in data_lists:
    df = pd.read_csv(data, sep="|", dtype={'std_ym': 'str', 'hcode': 'str', 'ry_s_cd': 'str'}, encoding="utf-8")
    df = df[(df['hcode'].str.startswith('26170')) & (df['card_pos'] == '개인사업자')].copy()
    res.append(df.groupby('std_ym', as_index=False)[['amt','cnt']].sum().sort_values('std_ym', ascending=False))

results = pd.concat(res, ignore_index=True)
results.to_csv(r"D:\02_사업관리\2026년\데이터 분석 컨설팅\20260402_해양수도정책과\\월별_동구_가맹점창폐업 현황.csv", sep=",",
               encoding="utf-8", index=False)
```
