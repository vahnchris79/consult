```python
import pandas as pd
import numpy as np
import PublicDataReader as pdr
```

```python
hcode = pdr.code_hdong()
hcode = hcode.loc[(hcode['시도코드']=='26') & (hcode['시군구코드']!='') & (hcode['읍면동명']!='') & (hcode['말소일자']=='')]
hcode = hcode[['시군구코드','시군구명','행정동코드', '읍면동명']]
hcode.columns = ['gcode', 'gname', 'hcode', 'hname']
inner1 = pd.read_csv("project/4_대상지_관내유입_생활인구_현황.csv", sep="|", dtype={'std_ym': str, 'id': str, 'inflow_cd': str}, encoding="utf-8")
inner1 = pd.merge(inner1, hcode, "inner", left_on="inflow_cd", right_on="hcode")
inner1.head()
```

```python
group = inner1.groupby(['std_ym','roadnm','gcode','gname','regists']).agg({'pop': np.sum}).reset_index().sort_values(by=['std_ym', 'gcode'])
group.to_csv("4_대상지_관내유입_생활인구_현황(합계).csv", sep="|", encoding="utf-8", index=False)
```
