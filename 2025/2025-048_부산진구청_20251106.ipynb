```python
import pandas as pd

path = "D:\\02_사업관리\\2025년\\데이터 분석 컨설팅\\20251106_부산진구청\\03_집계결과\\카드소비\\전포카페거리\\"
output = "D:\\02_사업관리\\2025년\\데이터 분석 컨설팅\\20251106_부산진구청\\"
rycode = pd.read_csv("D:\\02_사업관리\\코드자료\\bccd_sec_code.csv", sep="|", encoding="utf-8")

df1 = pd.read_csv(path + "bccd_d_sec_p_202410_jeonpocafe.csv", sep=",", dtype={'std_ymd': str, 'hcode': str, 'id': str},
                  encoding="utf-8")
df2 = pd.read_csv(path + "bccd_d_sec_p_202510_jeonpocafe.csv", sep=",", dtype={'std_ymd': str, 'hcode': str, 'id': str},
                  encoding="utf-8")

df3 = pd.concat([df1, df2], ignore_index=True)
df3['std_ymd'] = df3['std_ymd'].astype('datetime64[ns]')
df3['weekday'] = df3['std_ymd'].dt.day_name('ko_KR')
df3['std_ymd'] = df3['std_ymd'].astype(str)
df3 = pd.merge(df3, rycode, "left", "ry_m_cd")

group = round(df3.groupby(['std_ymd','weekday','ry_m_nm'], as_index=False)[['amt', 'cnt']].sum())
group.to_excel(output + "카드소비(전포카페거리).xlsx", engine="openpyxl", index=False)
```
