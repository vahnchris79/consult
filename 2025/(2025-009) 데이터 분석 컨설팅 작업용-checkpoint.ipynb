#### 최근 5년간 부산에서 발생한 화재발생 현황 및 변화추이

```python

# 요일정보 추출
fire['요일코드'] = fire['화재발생년월일'].dt.weekday 

# 8개 구군씩 분할하여 구군별 그래프의 시인성 확보
fire1 = fire[fire['시군구코드'].str.contains('26110|26140|26170|26200|26230|26260|26290|26320')].copy()
fire2 = fire[fire['시군구코드'].str.contains('26350|26380|26410|26440|26470|26500|26530|26710')].copy()

# 시각화 그래프 설정
fig, axes = plt.subplots(nrows=3, ncols=1, figsize=(18, 9))
fig.patch.set_alpha(0)  # 전체 배경 투명

# 화재발생(년)별 화재발생건수 집계
res_year = fire.groupby(['화재발생(년)','시군구코드','시군구'])[['화재건수']].sum().reset_index().sort_values(by=['화재발생(년)','시군구코드'])
sns.barplot(res_year, x='시군구', y='화재건수',  hue='화재발생(년)', errorbar=None, width=0.5, palette='Set1', ax=axes[0])
axes[0].set_facecolor("none")
axes[0].grid(color="gray", linestyle="--", linewidth=0.5)
axes[0].set_xlabel("시군구")
axes[0].set_ylabel("발생건수")
axes[0].legend(title='발생연도', loc='upper left', bbox_to_anchor=(1.0, 1.025))

# 화재발생(월)별 화재발생건수 집계
res_month = fire.groupby(['화재발생(년)', '화재발생(월)'])[['화재건수']].sum().reset_index().sort_values(by=['화재발생(년)','화재발생(월)'])
sns.barplot(res_month, x='화재발생(월)', y='화재건수', hue='화재발생(년)', errorbar=None, width=0.5, palette='Set1', ax=axes[1])
axes[1].set_facecolor("none")
axes[1].grid(color="gray", linestyle="--", linewidth=0.5)
axes[1].set_xlabel("화재발생(월)")
axes[1].set_ylabel("발생건수")
axes[1].legend(title='발생연도', loc='upper left', bbox_to_anchor=(1.0, 1.025))

# 화재발생(요일)별 화재발생건수 집계
res_dayname = fire.groupby(['화재발생(년)','요일','요일코드'])[['화재건수']].sum().reset_index().sort_values(by=['화재발생(년)','요일코드'])
sns.barplot(res_dayname, x='요일', y='화재건수', hue='화재발생(년)', errorbar=None, width=0.5, palette='Set1', ax=axes[2])
axes[2].set_facecolor("none")
axes[2].grid(color="gray", linestyle="--", linewidth=0.5)
axes[2].set_xlabel("요일")
axes[2].set_ylabel("발생건수")
axes[2].legend(title='발생연도', loc='upper left', bbox_to_anchor=(1.0, 1.025))

plt.tight_layout()
plt.show()
```

### 연도별 날씨, 온도, 습도에 따른 화재발생 현황 및 추이 분석

```python

# 그래프 설정
fig, axes = plt.subplots(nrows=3, ncols=1, figsize=(14, 8))
fig.patch.set_alpha(0)  # 전체 배경 투명

for i in fire.index:
    if fire.loc[i, '날씨'] == '비/눈':
        fire.loc[i, '날씨'] = '눈/비'

# 온도별 화재발생건수 집계, 10도 단위별로 묶어서 표시
fire.loc[(fire['온도(℃)'] >= -10.0) & (fire['온도(℃)'] <= -0.0), '온도범위'] ='-10도'
fire.loc[(fire['온도(℃)'] >= 0.0) & (fire['온도(℃)'] <= 9.0), '온도범위'] ='+09도'
fire.loc[(fire['온도(℃)'] >= 10.0) & (fire['온도(℃)'] <= 19.0), '온도범위'] ='+10도'
fire.loc[(fire['온도(℃)'] >= 20.0) & (fire['온도(℃)'] <= 29.0), '온도범위'] ='+20도'
fire.loc[(fire['온도(℃)'] >= 30.0) & (fire['온도(℃)'] <= 39.0), '온도범위'] ='+30도'

# 습도별 화재발생건수 집계, 10도 단위별로 묶어서 표시
fire.loc[(fire['습도'] >= 0.0) & (fire['습도'] <= 9.0), '습도범위'] ='09%'
fire.loc[(fire['습도'] >= 10.0) & (fire['습도'] <= 19.0), '습도범위'] ='10%'
fire.loc[(fire['습도'] >= 20.0) & (fire['습도'] <= 29.0), '습도범위'] ='20%'
fire.loc[(fire['습도'] >= 30.0) & (fire['습도'] <= 39.0), '습도범위'] ='30%'
fire.loc[(fire['습도'] >= 40.0) & (fire['습도'] <= 49.0), '습도범위'] ='40%'
fire.loc[(fire['습도'] >= 50.0) & (fire['습도'] <= 59.0), '습도범위'] ='50%'
fire.loc[(fire['습도'] >= 60.0) & (fire['습도'] <= 69.0), '습도범위'] ='60%'
fire.loc[(fire['습도'] >= 70.0) & (fire['습도'] <= 79.0), '습도범위'] ='70%'
fire.loc[(fire['습도'] >= 80.0) & (fire['습도'] <= 89.0), '습도범위'] ='80%'
fire.loc[(fire['습도'] >= 90.0) & (fire['습도'] <= 99.9), '습도범위'] ='90%'

# 닐씨별 화재발생건수 집계
res_weather = fire.groupby(['화재발생(년)', '날씨'])[['화재건수']].sum().reset_index().sort_values(by=['화재발생(년)', '날씨'])
sns.barplot(res_weather, x='날씨', y='화재건수', hue='화재발생(년)', errorbar=None, width=0.45, palette='Set1', ax=axes[0])
axes[0].set_facecolor("none")
axes[0].grid(color="gray", linestyle="--", linewidth=0.5)
axes[0].set_xlabel("날씨")
axes[0].set_ylabel("발생건수")
axes[0].legend(title='발생연도', loc='upper left', bbox_to_anchor=(0.995, 1.025))

res_temp = fire.groupby(['화재발생(년)', '온도범위'])[['화재건수']].sum().reset_index().sort_values(by=['화재발생(년)','온도범위'])
sns.barplot(res_temp, x='온도범위', y='화재건수', hue='화재발생(년)', errorbar=None, width=0.45, palette='Set1', ax=axes[1])
axes[1].set_facecolor("none")
axes[1].grid(color="gray", linestyle="--", linewidth=0.5)
axes[1].set_xlabel("온도")
axes[1].set_ylabel("발생건수")
axes[1].legend(title='발생연도', loc='upper left', bbox_to_anchor=(0.995, 1.025))

res_hum = fire.groupby(['화재발생(년)', '습도범위'])[['화재건수']].sum().reset_index().sort_values(by=['화재발생(년)', '습도범위'])
sns.barplot(res_hum, x='습도범위', y='화재건수', hue='화재발생(년)', errorbar=None, width=0.45, palette='Set1', ax=axes[2])
axes[2].set_facecolor("none")
axes[2].grid(color="gray", linestyle="--", linewidth=0.5)
axes[2].set_xlabel("습도")
axes[2].set_ylabel("발생건수")
axes[2].legend(title='발생연도', loc='upper left', bbox_to_anchor=(0.995, 1.025))

plt.tight_layout()
#plt.savefig("그래프/입찰금액_비교2_유찰.png", dpi=200, transparent=True)
plt.show()
```

```python
import warnings
warnings.filterwarnings('ignore')
import geopandas as gpd
import matplotlib.pyplot as plt
from datetime import datetime

grid050 = gpd.read_file("2025_009_데이터 분석 컨설팅_20250227.gpkg", layer="bigdata_50m_busan", driver="GPKG", encoding="utf-8")
grid050['x'] = grid050.geometry.centroid.x
grid050['y'] = grid050.geometry.centroid.y
grid100 = gpd.read_file("2025_009_데이터 분석 컨설팅_20250227.gpkg", layer="ngii_100m_busan", driver="GPKG", encoding="utf-8")

older = pd.read_csv("집계자료/월별_구군_50m_고령인구.csv", dtype={'gcode': str, 'id': str},  sep=",", encoding="utf-8")
older = pd.merge(older, grid050, how="left", on="id")
golder = gpd.GeoDataFrame(older, geometry=gpd.points_from_xy(older['x'], older['y']), crs="EPSG:5179")
golder = gpd.sjoin(golder, grid100, how="inner", predicate="intersects")
golder['화재발생(년)'] = golder['std_ym'].str[:4]
golder = golder.groupby(['화재발생(년)', 'gcode'])[['avg60', 'avg70']].sum().reset_index().sort_values(by=['화재발생(년)', 'gcode'])
golder.columns = ['화재발생(년)', '시군구코드', '60대', '70대']
#golder = golder.melt(id_vars=['화재발생(년)','시군구코드'], value_vars=['60대', '70대'], var_name='연령대', value_name='인구수')
golder = pd.merge(golder, gcode, how="left", on='시군구코드')

# 화재발생 현황자료의 시도~도로명주소를 조합하여 좌표로 바꾼 후 다시 데이터프레임으로 변환
cause = fire[['화재발생년월일','시도','시군구코드','시군구','읍면동','지번동','지번리','세부주소','도로명주소','발화요인대분류','사망']]
for i in cause.index:
    if cause.loc[i, '시군구'] != '기장군' and cause.loc[i, '세부주소'] != '' or cause.loc[i, '도로명주소'] == '':
        cause.loc[i, '주소'] = cause.loc[i, '시도'] + ' ' + cause.loc[i, '시군구'] + ' ' + \
        str(cause.loc[i, '지번동']) + ' ' + str(cause.loc[i, '세부주소']).split(' ')[0]
    elif cause.loc[i, '시군구'] == '기장군' and cause.loc[i, '세부주소'] != '' or cause.loc[i, '도로명주소'] == '':
        cause.loc[i, '주소'] = cause.loc[i, '시도'] + ' ' + cause.loc[i, '시군구'] + ' ' + \
        str(cause.loc[i, '지번동']) + ' ' + str(cause.loc[i, '지번리']) + ' ' + str(cause.loc[i, '세부주소']).split(' ')[0]        
    elif cause.loc[i, '시군구'] == '기장군' and cause.loc[i, '세부주소'] != '' or cause.loc[i, '도로명주소'] != '':
        cause.loc[i, '주소'] = cause.loc[i, '시도'] + ' ' + cause.loc[i, '시군구'] + ' ' + \
        str(cause.loc[i, '지번동']) + ' ' + str(cause.loc[i, '지번리']) + ' ' + str(cause.loc[i, '도로명주소'])

coords = pd.read_excel("집계자료/화재발생좌표.xlsx", usecols=['주소', '위도', '경도'], engine="openpyxl")
cause = pd.merge(cause, coords, how="left", on="주소")
cause = gpd.GeoDataFrame(cause, geometry=gpd.points_from_xy(x=cause['경도'], y=cause['위도'], crs="EPSG:4326")).to_crs(5179)

# 100m격자ID를 추가하고 사망자 수 집계
cause = gpd.sjoin(cause, grid100, predicate='within')
cause['화재발생(년)'] = fire['화재발생년월일'].dt.strftime("%Y")
group = cause.groupby(['화재발생(년)', '시군구코드', '시군구', '발화요인대분류'])[['사망']].sum().round(0).reset_index().sort_values(by=['화재발생(년)','시군구코드'])

# 노후주택 집계
buld = gpd.read_file("2025_009_데이터 분석 컨설팅_20250227.gpkg", layer="GIS건물통합정보", driver="GPKG").dropna(subset=['사용승인일자'], axis=0)
buld['건물수'] = buld['원천도형ID']
buld['사용승인연도'] = [int(buld.loc[i, '사용승인일자'].split('-')[0]) for i in buld.index]
buld['건물연령'] = 2025 - buld['사용승인연도']
buld = buld[buld['건물연령']>=30].copy()
for i in buld.index:
    if buld.loc[i, '건물연령'] >= 30 and buld.loc[i, '건물연령'] <= 39:
        buld.loc[i, '건물연령대'] = '30년이상 40년미만'
    elif buld.loc[i, '건물연령'] >= 40 and buld.loc[i, '건물연령'] <= 49:
        buld.loc[i, '건물연령대'] = '40년이상 50년미만'
    elif buld.loc[i, '건물연령'] >= 50 and buld.loc[i, '건물연령'] <= 59:
        buld.loc[i, '건물연령대'] = '50년이상 60년미만'
    elif buld.loc[i, '건물연령'] >= 60 and buld.loc[i, '건물연령'] <= 69:
        buld.loc[i, '건물연령대'] = '60년이상 70년미만'
    elif buld.loc[i, '건물연령'] >= 70 and buld.loc[i, '건물연령'] <= 79:
        buld.loc[i, '건물연령대'] = '70년이상 80년미만'
    elif buld.loc[i, '건물연령'] >= 80 and buld.loc[i, '건물연령'] <= 89:
        buld.loc[i, '건물연령대'] = '80년이상 90년미만'
    elif buld.loc[i, '건물연령'] >= 90 and buld.loc[i, '건물연령'] < 100:
        buld.loc[i, '건물연령대'] = '90년이상 100년미만'
    else:
        buld.loc[i, '건물연령대'] = '100년이상'
buld['시군구코드'] = [buld.loc[i, '법정동코드'][:5] for i in buld.index]
buld = pd.merge(buld, gcode, how='inner', on='시군구코드')
gbuld = buld.groupby(['시군구코드', '시군구', '건물연령대'])[['건물수']].count().reset_index().sort_values(by='시군구코드')

# 연도별 구군별 인구밀집 현황
ratio = pd.read_csv("집계자료/연도별 구군별 인구밀도.csv", sep=",", dtype={'std_year': str, 'gcode': str, '': str}, 
                    header=0, names=['화재발생(년)', '시군구코드', '시군구', '서비스인구', '면적', '인구밀도'], encoding="utf-8")

# 시각화 그래프 설정: 시인성 확보를 위해 월별과 요일별의 캔버스 사이즈 별도 조정
fig, axes = plt.subplots(nrows=6, ncols=1,figsize=(18, 15))
fig.patch.set_alpha(0)  # 전체 배경 투명

sns.barplot(group, x='시군구', y='사망', hue='화재발생(년)', errorbar=None, width=0.8, palette='Set1', ax=axes[0])
axes[0].set_facecolor("none")
axes[0].grid(color="gray", linestyle="--", linewidth=0.6)
axes[0].set_xlabel('시군구')
axes[0].set_ylabel('사망자(명)')
axes[0].legend(title='발생연도', loc='upper left', bbox_to_anchor=(1.00, 1.025))

sns.barplot(group, x='발화요인대분류', y='사망', hue='화재발생(년)', errorbar=None, width=0.8, palette='Set1', ax=axes[1])
axes[1].set_facecolor("none")
axes[1].grid(color="gray", linestyle="--", linewidth=0.6)
axes[1].set_xlabel('발화요인')
axes[1].set_ylabel('사망자(명)')
axes[1].legend(title='발생연도', loc='upper left', bbox_to_anchor=(1.00, 1.025))

sns.barplot(golder, x='시군구', y='60대', hue='화재발생(년)', errorbar=None, width=0.8, palette='Set1', ax=axes[2])
axes[2].set_facecolor("none")
axes[2].grid(color="gray", linestyle="--", linewidth=0.6)
axes[2].set_xlabel('시군구')
axes[2].set_ylabel('60대인구')
axes[2].legend(title='발생연도', loc='upper left', bbox_to_anchor=(1.00, 1.025))

sns.barplot(golder, x='시군구', y='70대', hue='화재발생(년)', errorbar=None, width=0.8, palette='Set1', ax=axes[3])
axes[3].set_facecolor("none")
axes[3].grid(color="gray", linestyle="--", linewidth=0.6)
axes[3].set_xlabel('시군구')
axes[3].set_ylabel('70대인구')
axes[3].legend(title='발생연도', loc='upper left', bbox_to_anchor=(1.00, 1.025))

sns.barplot(gbuld, x='시군구', y='건물수', hue='건물연령대', errorbar=None, width=0.8, palette='OrRd', ax=axes[4])
axes[4].set_facecolor("none")
axes[4].grid(color="gray", linestyle="--", linewidth=0.6)
axes[4].set_xlabel('시군구')
axes[4].set_ylabel('건물연령대')
axes[4].legend(title='건물연령', loc='upper left', bbox_to_anchor=(1.00, 1.025))

sns.barplot(ratio, x='시군구', y='인구밀도', hue='화재발생(년)', errorbar=None, width=0.8, palette='Reds', ax=axes[5])
axes[5].set_facecolor("none")
axes[5].grid(color="gray", linestyle="--", linewidth=0.6)
axes[5].set_xlabel('시군구')
axes[5].set_ylabel('인구밀도')
axes[5].legend(title='발생연도', loc='upper left', bbox_to_anchor=(1.00, 1.025))

plt.tight_layout()
#plt.savefig("그래프/입찰금액_비교2_유찰.png", dpi=200, transparent=True)
plt.show()
```
