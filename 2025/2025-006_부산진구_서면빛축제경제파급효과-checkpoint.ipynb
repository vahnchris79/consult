```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta

plt.rcParams['font.family'] = 'NanumGothic'
plt.rcParams['font.size'] = 11
plt.rcParams['axes.unicode_minus'] = False

# 그래프 설정
fig, axes = plt.subplots(2, 2, figsize=(14, 7))
palette = {0: '#2ca02c', 1: '#d62728'}

# 기온데이터: 기가 전, 기간 중 구분컬럼 추가
before_temp = pd.read_csv("분석용데이터/기온강수데이터/ta_20250212113637.csv", skiprows=7, encoding="cp949").fillna(0)
before_temp['festival_peroid'] = 0
during_temp = pd.read_csv("분석용데이터/기온강수데이터/ta_20250212113718.csv", skiprows=7, encoding="cp949").fillna(0)
during_temp['festival_peroid'] = 1

# 강수데이터: 기가 전, 기간 중 구분컬럼 추가
before_rain = pd.read_csv("분석용데이터/기온강수데이터/rn_20250212113856.csv", skiprows=7, encoding="cp949").fillna(0)
before_rain['festival_peroid'] = 0
during_rain = pd.read_csv("분석용데이터/기온강수데이터/rn_20250212113926.csv", skiprows=7, encoding="cp949").fillna(0)
during_rain['festival_peroid'] = 1

# 데이터 합치기
weather = pd.concat([before_temp, during_temp], ignore_index=True)
rain = pd.concat([before_rain, during_rain], ignore_index=True)

# 기온데이터 전처리
weather['std_ymd'] = [pd.to_datetime(weather.loc[i, '날짜'].replace("\t", ""), format="%Y-%m-%d") for i in weather.index]
weather['weekday'] = weather['std_ymd'].dt.dayofweek
weekday_list = ['월', '화', '수', '목', '금', '토', '일']
weather['요일'] = weather.apply(lambda x: weekday_list[x['weekday']], axis=1)
weather['weeknum'] = ((weather['std_ymd'].dt.day - 1) // 7 + 1).astype(str)
weather['주별'] = weather['weeknum'] + '주'
weather1 = weather[['std_ymd', '요일', '주별', 'festival_peroid', '평균기온(℃)','최저기온(℃)','최고기온(℃)']]
weather1.columns = ['std_ymd', '요일', '주별', 'festival_peroid', 'avg_temp', 'min_temp', 'max_temp']

# 강수데이터 전처리
rain['std_ymd'] = pd.to_datetime(rain['날짜'], format="%Y-%m-%d")
rain['weekday'] = rain['std_ymd'].dt.dayofweek
weekday_list = ['월', '화', '수', '목', '금', '토', '일']
rain['요일'] = rain.apply(lambda x: weekday_list[x['weekday']], axis=1)
rain['weeknum'] = ((rain['std_ymd'].dt.day - 1) // 7 + 1).astype(str)
rain['주별'] = rain['weeknum'] + '주'
rain1 = rain[['std_ymd', '요일', '주별', 'festival_peroid', '강수량(mm)']]
rain1.columns = ['std_ymd', '요일', '주별', 'festival_peroid', 'rain_height']

# 기온, 강수데이터 합치기
df2 = pd.merge(weather1, rain1, how="inner", on=['std_ymd','요일','주별','festival_peroid'])

# 요일별 기간 전과 중의 평균기온
sns.barplot(df2, x='요일', y='avg_temp', hue='festival_peroid', errorbar=('ci', 0), palette=palette, width=0.25, ax=axes[0][0])
axes[0][0].set_facecolor('none')
axes[0][0].grid(color='gray', linestyle='--', linewidth=0.5)
axes[0][0].set_xlabel('요일별')
axes[0][0].set_ylabel('평균기온(℃)')
axes[0][0].legend([], [], frameon=False)
sns.barplot(df2, x='요일', y='rain_height', hue='festival_peroid', errorbar=('ci', 0), palette=palette, width=0.25, ax=axes[0][1])
axes[0][1].set_facecolor('none')
axes[0][1].grid(color='gray', linestyle='--', linewidth=0.5)
axes[0][1].set_xlabel('요일별')
axes[0][1].set_ylabel('강수량(mm)')
handles, _ = axes[0][1].get_legend_handles_labels()
axes[0][1].legend(handles=handles, labels=['기간전', '기간중'], loc="upper right")

sns.barplot(df2, x='주별', y='avg_temp', hue='festival_peroid', errorbar=('ci', 0), palette=palette, width=0.25, ax=axes[1][0])
axes[1][0].set_facecolor('none')
axes[1][0].grid(color='gray', linestyle='--', linewidth=0.5)
axes[1][0].set_xlabel('주차별')
axes[1][0].set_ylabel('평균기온(℃)')
axes[1][0].legend([], [], frameon=False)
sns.barplot(df2, x='주별', y='rain_height', hue='festival_peroid', errorbar=('ci', 0), palette=palette, width=0.25, ax=axes[1][1])
axes[1][1].set_facecolor('none')
axes[1][1].grid(color='gray', linestyle='--', linewidth=0.5)
axes[1][1].set_xlabel('주차별')
axes[1][1].set_ylabel('강우량(mm)')
handles, _ = axes[1][1].get_legend_handles_labels()
axes[1][1].legend(handles=handles, labels=['기간전', '기간중'], loc="upper right")

plt.tight_layout()
plt.savefig("(시각화)시계열분석4.png", dpi=200)
plt.show()
```

```python
# 축제 전과 축제 중 생활인구 및 이용금액 평균과 표준편차 계산
pop_before = df[df["festival_peroid"] == 0]["생활인구(명)"].dropna()
pop_during = df[df["festival_peroid"] == 1]["생활인구(명)"].dropna()

amt_before = df[df["festival_peroid"] == 0]["이용금액(만원)"].dropna()
amt_during = df[df["festival_peroid"] == 1]["이용금액(만원)"].dropna()

# 평균 및 표준편차 계산
stats_summary = pd.DataFrame({
    "Metric": ["생활인구(명)", "이용금액(만원)"],
    "축제기간 전 - 평균": [pop_before.mean(), amt_before.mean()],
    "축제기간 전 - 표준편차": [pop_before.std(), amt_before.std()],
    "축제기간 중 - 평균": [pop_during.mean(), amt_during.mean()],
    "축제기간 중 - 표준편차": [pop_during.std(), amt_during.std()],
})

# T-검정 수행
from scipy import stats

t_stat_pop, p_value_pop = stats.ttest_ind(pop_before, pop_during, equal_var=False)
t_stat_amt, p_value_amt = stats.ttest_ind(amt_before, amt_during, equal_var=False)

# T-검정 결과 추가
stats_summary["T-Statistic"] = [t_stat_pop, t_stat_amt]
stats_summary["P-Value"] = [p_value_pop, p_value_amt]

stats_summary
```

### 기간 전과 기간 중 날씨요인에 따른 생활인구와 소비변화 분석: 다중회귀분석

```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.tools import add_constant

# 생활인구, 카드이용 컬럼 정리
df1 = df[['std_ym', 'weekday', 'weeknum', 'festival_peroid', '생활인구(명)', '이용금액(만원)']]

# 기온과 강수량 컬럼 정리
weather2 = weather[['std_ymd', 'weekday', 'weeknum', 'festival_peroid', '평균기온(℃)']]
rain2 = rain[['std_ymd', 'weekday', 'weeknum', 'festival_peroid', '강수량(mm)']]

# 분석용 데이터셋 생성
df3 = pd.merge(weather2, rain2, how="inner", on=['std_ymd', 'weekday', 'weeknum', 'festival_peroid'])
df3['std_ym'] = df3['std_ymd'].astype(str).str[:7]
result = pd.merge(df1, df3, how="inner", on=['std_ym', 'weekday', 'weeknum', 'festival_peroid']).drop(columns='std_ymd')
result.columns=['std_ym','weekday','weeknum','festival_peroid','population','cardamt','avgtemp','rainheight']
#datasets.to_csv("분석용데이터/datasets_v1.csv", encoding="utf-8", index=False)

# 다중회귀분석
# 1. 생활인구를 종속변수로 하는 경우
X_pop = result[['avgtemp', 'rainheight', 'festival_peroid']]
y_pop = result['population']

# 상수항 추가
X_pop = add_constant(X_pop)

# OLS 회귀모형 적합
model_pop = sm.OLS(y_pop, X_pop).fit()

print("\n===== 생활인구에 대한 다중회귀분석 결과 =====")
print(model_pop.summary())

# 2. 소비금액에 대한 다중회귀분석
X_amt = result[['avgtemp', 'rainheight', 'festival_peroid']]
y_amt = result['cardamt']

# 상수항 추가
X_amt = add_constant(X_amt)

# OLS 회귀모형 분석
model_amt = sm.OLS(y_amt, X_amt).fit()

print("\n===== 이용금액에 대한 다중회귀분석 결과 =====")
print(model_amt.summary())
```
