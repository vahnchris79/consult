```python
import pandas as pd
import glob

result = []
data_lists = glob.glob("D:/Project/datacheck/02_carduse/biz_emd/*.txt", recursive=True)
for data in data_lists:
    df = pd.read_csv(data, sep="|", dtype={'std_ym': str, 'hcode': str}, encoding="utf-8")
    result.append(df[df['hcode'].str.startswith('26230')].copy())

merged = pd.concat(result, ignore_index=True)    
merged.to_csv("D:/10_분석업무/05_구군요청/26230_부산진구/2025년/20250217/토지정보과/최종결과/busan_biz_emd_26230.csv", 
              sep=",", encoding="utf-8", index=False)
```
